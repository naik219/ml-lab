```python
# %pip install hmmlearn
import numpy as np
from hmmlearn import hmm
```

```python
def index(V, O):
    arr = []
    for i in range(len(O)):
        for j in range(len(V)):
            if O[i] == V[j]:
                arr.append(j)
    return arr
```

```python
# S = input("Enter the Hidden states: ").split()
S = ["Walking", "Running", "Resting"]

# V = input("Enter the Observation Symbols: ").split()
V = ["Low", "Medium", "High"]

# pi = [int(x) for x in input("Enter the intial state distribution: ").split()]
pi = [0.5, 0.3, 0.2]

# print("\nEnter transition matrix:")
#A = []
#for i in range(N):
#    row = list(map(float, input().split()))
#    A.append(row)
#A = np.array(A)

A = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.3, 0.2, 0.5]
])

# Emission matrix
#print("\nEnter emission matrix:")
#B = []
#for i in range(N):
#    row = list(map(float, input().split()))
#    B.append(row)
#B = np.array(B)

B = np.array([
    [0.6, 0.3, 0.1],
    [0.1, 0.3, 0.6],
    [0.7, 0.2, 0.1]
])

# O = input("Enter the Observation Sequence: ").split()
O = ["High", "Medium", "Low", "High"]

# Convert observations to indices
O_index = np.array(index(V, O)).reshape(-1, 1)

## Hidden Marko Model
model = hmm.CategoricalHMM(
    n_components = len(S), ## No of Hidden States
    init_params = ""
)

model.startprob_ = pi
model.transmat_ = A
model.emissionprob_ = B

log_prob = model.score(O_index)

print("Observation Sequence:", O)
print("Log Likelihood:", log_prob)
print("Likelihood:", np.exp(log_prob))
```

```python
sequences = {
    "O1": ["Low", "Medium", "High", "Medium"],
    "O2": ["High", "High", "High", "High"],
    "O3": ["Low", "Low", "Low", "Low"],
    "O4": ["Medium", "Medium", "Medium", "Medium"]
}

for name, seq in sequences.items():

    O_index = np.array(index(V, seq)).reshape(-1, 1)

    log_prob = model.score(O_index)

    print(name)
    print("Sequence:", seq)
    print("Log Likelihood:", log_prob)
    print("Likelihood:", np.exp(log_prob))
    print()
```
