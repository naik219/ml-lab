```python
import numpy as np
```

```python
def index(V, O):
    arr = []
    for i in range(len(O)):
        for j in range(len(V)):
            if O[i] == V[j]:
                arr.append(j)
    return arr
```

```python
def for_alg(S, A, B, pi, V, O):
    
    row, column = len(O), len(S) # 4, 3
    ## dm = [[0 for _ in range(column)] for _ in range(row)]
    dm = np.zeros((row, column))
    
    # Intialization
    for j in range(column):
        dm[0][j] = pi[j] * B[j][O[0]]

    ## Dynamic Programmin
    for i in range(1, row):
        for j in range(column):
            for k in range(column):
                dm[i][j] += dm[i-1][k] * A[k][j] * B[j][O[i]]

    return sum(dm[row - 1])
```

```python
# arr = [int(x) for x in input("Enter the array: ").split()]

# S = input("Enter the Hidden states: ").split()
S = ["Walking", "Running", "Resting"]

# V = input("Enter the Observation Symbols: ").split()
V = ["Low", "Medium", "High"]

# pi = [int(x) for x in input("Enter the intial state distribution: ").split()]
pi = [0.5, 0.3, 0.2]

# A = [[[int(x) for x in input("Enter the Probability: ").split()]] for _ in range(len(V))]
A = [
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.3, 0.2, 0.5]
]

B = [
    [0.6, 0.3, 0.1],
    [0.1, 0.3, 0.6],
    [0.7, 0.2, 0.1]
]

# O = input("Enter the Observation Sequence: ").split()
O = ["High", "Medium", "Low", "High"]
O_index = index(V, O)
print(O_index)

print("The likelihood is: ", for_alg(S, A, B, pi, V, O_index))
```
